# Phase 2.2: Feature Engineering
This notebook demonstrates feature engineering for the Predictive Maintenance system.


## Step 1: Load Data & Initialize
We load the raw sensor data and run the full feature pipeline.


```python
import pandas as pd
from src.data_pipeline.ingestion import load_raw_data
from src.data_pipeline.feature_store import run_full_feature_pipeline, select_features_by_importance
from src.data_pipeline.preprocessing import split_data

df = load_raw_data('data/raw/ai4i2020.csv')
df_feat = run_full_feature_pipeline(df)
print('DataFrame columns after feature engineering:')
print(df_feat.columns.tolist())
```

## Step 2: Feature Selection by Importance
We use a Random Forest Classifier to identify the most predictive features.


```python
X = df_feat.drop(columns=['Machine failure', 'TWF', 'HDF', 'PWF', 'OSF', 'RNF'])
y = df_feat['Machine failure']
X_train, X_val, X_test, y_train, y_val, y_test = split_data(
    pd.concat([X, y], axis=1), 'Machine failure'
)
top_features = select_features_by_importance(X_train, y_train, n_features=8)
print('Top 8 features selected by Random Forest Importance:')
print(top_features)
```

## Step 3: Save Engineered Features
We persist the engineered feature matrix for model training.


```python
import os
os.makedirs('data/features', exist_ok=True)
df_feat.to_csv('data/features/engineered_features.csv', index=False)
print('Engineered features saved to data/features/engineered_features.csv')
```
